# Quarterback Analysis

### Vorbereitungen

##### Pakete laden

```python
import nflreadpy as nfl
import polars as pl
import numpy as np
```

##### Datengrundlage

```python
seasons = np.arange(2003, 2026).tolist() # 2003 - 2025
```

```python
(
    nfl.load_player_stats(seasons)
    .filter(
        (pl.col('player_display_name') == 'Josh Allen') & (pl.col('position') == "QB"))
    .select([
        'player_id', 'player_display_name', 'position', 'position_group',
        'season', 'week', 'season_type', 'team', 'opponent_team',
        'completions', 'attempts', 'passing_yards', 'passing_tds', 'passing_interceptions',
        'sacks_suffered', 'sack_yards_lost', 'sack_fumbles', 'sack_fumbles_lost',
        'passing_air_yards', 'passing_yards_after_catch', 'passing_first_downs', 'passing_epa', 'passing_cpoe', 'passing_2pt_conversions',
        'pacr',
        'carries', 'rushing_yards', 'rushing_tds', 'rushing_fumbles', 'rushing_fumbles_lost', 'rushing_first_downs', 'rushing_epa', 'rushing_2pt_conversions', 'racr',
        'target_share', 'air_yards_share', 'wopr', 'special_teams_tds','misc_yards',
        'penalties', 'penalty_yards', 'fantasy_points', 'fantasy_points_ppr'
    ])
)
```
